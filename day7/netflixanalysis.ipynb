```python
missing_values = df.isnull().sum()

print(missing_values)
```

```python
df_clean = df.copy()
```

```python
df_clean.isnull().sum()
```

```python
df_clean["director"] = df_clean["director"].fillna("Unknown")
df_clean["cast"] = df_clean["cast"].fillna("Unknown")
df_clean["country"] = df_clean["country"].fillna("Unknown")
df_clean["rating"] = df_clean["rating"].fillna("Unknown")
```

```python
df_clean["date_added"] = pd.to_datetime(
    df_clean["date_added"],
    errors="coerce"
)
```

```python
df_clean.isnull().sum()
```

```python
df_clean = df_clean.drop_duplicates()

print("Shape after removing duplicates:", df_clean.shape)
```

```python
df_clean["year_added"] = df_clean["date_added"].dt.year
df_clean["month_added"] = df_clean["date_added"].dt.month_name()

df_clean[["title", "date_added", "year_added", "month_added"]].head()
```

```python
df_clean["date_added"] = pd.to_datetime(
    df_clean["date_added"],
    format="mixed",
    errors="coerce"
)

print("Missing date_added values:")
print(df_clean["date_added"].isnull().sum())
```

```python
df_clean["year_added"] = df_clean["date_added"].dt.year
df_clean["month_added"] = df_clean["date_added"].dt.month_name()
```

```python
df_clean[["title", "date_added", "year_added", "month_added"]].head()
```

```python
print("Final dataset shape:", df_clean.shape)

print("\nRemaining missing values:")
print(df_clean.isnull().sum())
```

```python
type_counts = df_clean["type"].value_counts()

print("Content Type Distribution:")
print(type_counts)
```

```python
plt.figure(figsize=(8, 5))

type_counts.plot(
    kind="bar",
    edgecolor="black"
)

plt.title("Netflix Movies vs TV Shows")
plt.xlabel("Content Type")
plt.ylabel("Number of Titles")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
release_year_counts = df_clean["release_year"].value_counts().sort_index()

print(release_year_counts.tail(10))
```

```python
plt.figure(figsize=(12, 6))

plt.plot(
    release_year_counts.index,
    release_year_counts.values,
    marker="o"
)

plt.title("Netflix Content by Release Year")
plt.xlabel("Release Year")
plt.ylabel("Number of Titles")
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
country_data = df_clean["country"].dropna().str.split(", ")

country_counts = country_data.explode().value_counts()

print("Top 10 Countries:")
print(country_counts.head(10))
```

```python
plt.figure(figsize=(10, 6))

country_counts.head(10).sort_values().plot(
    kind="barh",
    edgecolor="black"
)

plt.title("Top 10 Countries by Number of Netflix Titles")
plt.xlabel("Number of Titles")
plt.ylabel("Country")
plt.tight_layout()
plt.show()
```

```python
genre_data = df_clean["listed_in"].dropna().str.split(", ")

genre_counts = genre_data.explode().value_counts()

print("Top 10 Genres:")
print(genre_counts.head(10))
```

```python
plt.figure(figsize=(10, 6))

genre_counts.head(10).sort_values().plot(
    kind="barh",
    edgecolor="black"
)

plt.title("Top 10 Netflix Genres")
plt.xlabel("Number of Titles")
plt.ylabel("Genre")
plt.tight_layout()
plt.show()
```

```python
rating_counts = df_clean["rating"].value_counts()

print("Netflix Ratings:")
print(rating_counts)
```

```python
plt.figure(figsize=(10, 6))

rating_counts.head(10).plot(
    kind="bar",
    edgecolor="black"
)

plt.title("Most Common Netflix Ratings")
plt.xlabel("Rating")
plt.ylabel("Number of Titles")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
movies = df_clean[df_clean["type"] == "Movie"].copy()

movies["duration_minutes"] = (
    movies["duration"]
    .str.replace(" min", "", regex=False)
    .astype(float)
)

movies[["title", "duration", "duration_minutes"]].head()
```

```python
print("Average Movie Duration:",
      round(movies["duration_minutes"].mean(), 2),
      "minutes")
```

```python
print("Shortest Movie:",
      movies["duration_minutes"].min(),
      "minutes")

print("Longest Movie:",
      movies["duration_minutes"].max(),
      "minutes")
```

```python
plt.figure(figsize=(10, 6))

plt.hist(
    movies["duration_minutes"].dropna(),
    bins=20,
    edgecolor="black"
)

plt.title("Distribution of Netflix Movie Durations")
plt.xlabel("Duration (minutes)")
plt.ylabel("Number of Movies")
plt.tight_layout()
plt.show()
```

```python
print("===== KEY ANALYSIS RESULTS =====")

print("\n1. Content Types:")
print(df_clean["type"].value_counts())

print("\n2. Most Common Rating:")
print(df_clean["rating"].value_counts().head(5))

print("\n3. Top 5 Countries:")
print(country_counts.head(5))

print("\n4. Top 5 Genres:")
print(genre_counts.head(5))

print("\n5. Average Movie Duration:")
print(round(movies["duration_minutes"].mean(), 2), "minutes")

print("\n6. Most Common Release Years:")
print(df_clean["release_year"].value_counts().head(5))

print("\n7. Earliest Release Year:")
print(df_clean["release_year"].min())

print("\n8. Latest Release Year:")
print(df_clean["release_year"].max())
```

```python
movie_count = (df_clean["type"] == "Movie").sum()
tv_count = (df_clean["type"] == "TV Show").sum()

movie_percentage = (movie_count / len(df_clean)) * 100

top_country = country_counts.index[0]
top_country_count = country_counts.iloc[0]

top_genre = genre_counts.index[0]
top_genre_count = genre_counts.iloc[0]

top_rating = df_clean["rating"].value_counts().index[0]
top_rating_count = df_clean["rating"].value_counts().iloc[0]

most_common_release_year = df_clean["release_year"].value_counts().index[0]
most_common_release_count = df_clean["release_year"].value_counts().iloc[0]

average_duration = movies["duration_minutes"].mean()

print("===== PROJECT INSIGHTS =====")

print(
    f"1. Netflix has {movie_count} Movies and {tv_count} TV Shows, "
    f"with Movies making up {movie_percentage:.2f}% of the dataset."
)

print(
    f"2. {top_country} is the country with the highest number of "
    f"Netflix titles represented in the dataset, with {top_country_count} titles."
)

print(
    f"3. {top_genre} is the most common genre/category, "
    f"appearing in {top_genre_count} titles."
)

print(
    f"4. {top_rating} is the most common content rating, "
    f"with {top_rating_count} titles."
)

print(
    f"5. {most_common_release_year} is the release year with the "
    f"highest number of titles, with {most_common_release_count} titles."
)

print(
    f"6. The average duration of Netflix Movies in the dataset "
    f"is approximately {average_duration:.2f} minutes."
)
```
