Import Libraries

```python
import torch

from transformers import (
    pipeline,
    AutoTokenizer,
    AutoModelForSeq2SeqLM
)

print("PyTorch version:", torch.__version__)
print("Transformers imported successfully.")
```

Understand the LLM Workflow

```python
print("DAY 15 — LARGE LANGUAGE MODELS")
print("=" * 40)

print("Traditional ML")
print("Data -> Features -> ML Model -> Prediction")

print("\nGenerative AI / LLM")
print("Prompt -> Tokenization -> Pretrained Model -> Generated Output")

print("\nToday we will explore:")
print("1. Text Generation")
print("2. Sentiment Analysis")
print("3. Text Summarization")
```

Test Text Generation

```python
generator = pipeline(
    "text-generation",
    model="openai-community/gpt2"
)

prompt = "Artificial intelligence is changing education because"

generation_output = generator(
    prompt,
    max_new_tokens=40,
    do_sample=False
)

print("Prompt:")
print(prompt)

print("\nGenerated Text:")
print(generation_output[0]["generated_text"])
```

Test Sentiment Analysis

```python
sentiment_model = pipeline(
    "sentiment-analysis",
    model="distilbert/distilbert-base-uncased-finetuned-sst-2-english"
)

sentiment_text = (
    "I really enjoyed this course because the explanations "
    "were clear, practical and easy to understand."
)

sentiment_output = sentiment_model(sentiment_text)

print("Input:")
print(sentiment_text)

print("\nSentiment Result:")
print(sentiment_output)
```

Test Multiple Sentiment Inputs

```python
sentiment_inputs = [
    "The project was excellent and I learned a lot.",
    "The explanation was confusing and the experience was disappointing.",
    "The course was useful and practical."
]

for text in sentiment_inputs:
    result = sentiment_model(text)

    print("Text:", text)
    print("Prediction:", result[0]["label"])
    print("Confidence:", round(result[0]["score"], 4))
    print()
```

Load Summarization Model

```python
summarizer_tokenizer = AutoTokenizer.from_pretrained(
    "sshleifer/distilbart-cnn-12-6"
)

summarizer_model = AutoModelForSeq2SeqLM.from_pretrained(
    "sshleifer/distilbart-cnn-12-6"
)

print("Summarization model loaded successfully.")
```

Summarize Text

```python
text_to_summarize = """
Artificial intelligence is becoming an important technology in many industries.
Machine learning allows computers to learn patterns from data, while generative
AI models can create text, images and other content. Large Language Models are
trained on large amounts of text and can perform tasks such as text generation,
summarization, question answering and translation. Hugging Face provides access
to many pretrained models that developers can use for these applications.
"""

inputs = summarizer_tokenizer(
    text_to_summarize,
    return_tensors="pt",
    truncation=True,
    max_length=1024
)

summary_ids = summarizer_model.generate(
    **inputs,
    max_length=80,
    min_length=25,
    num_beams=4,
    early_stopping=True
)

summary = summarizer_tokenizer.decode(
    summary_ids[0],
    skip_special_tokens=True
)

print("Original Text:")
print(text_to_summarize)

print("\nSummary:")
print(summary)
```

Compare the Three Tasks

```python
comparison = {
    "Text Generation": generation_output[0]["generated_text"],
    "Sentiment Analysis": sentiment_output[0],
    "Summarization": summary
}

for task, output in comparison.items():
    print("=" * 50)
    print(task)
    print("=" * 50)
    print(output)
    print()
```
